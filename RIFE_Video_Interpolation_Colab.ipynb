# 2. Interpolate video

- Upload your video & set the video path
- Adjust the `exp` slider, which controls the FPS of the output video
- The interpolated video will be saved in the same directory as original video

```
import os
import glob

video_path = "/content/inid2.mp4" #@param {type:"string"}
exp = 3 # @param {type:"slider", min:1, max:10, step:1}

# Auto-setup if the directory does not exist
if not os.path.exists('/content/arXiv2020-RIFE'):
    print("arXiv2020-RIFE not found. Performing quick setup...")
    !git clone https://github.com/hzwer/arXiv2020-RIFE /content/arXiv2020-RIFE
    !mkdir -p /content/arXiv2020-RIFE/train_log
    %cd /content/arXiv2020-RIFE/train_log
    !gdown 1APIzVeI-4ZZCEuIRE1m6WYfSCaOsi_7_
    !7z e RIFE_trained_model_v3.6.zip
    %cd /content/arXiv2020-RIFE/
    !gdown 1i3xlKb7ax7Y70khcTcuePi6E7crO_dFc
    !pip install scikit-video

# Patch the scikit-video library to replace np.float and np.int with builtins
try:
    import skvideo.io.abstract
    abstract_file = skvideo.io.abstract.__file__
    with open(abstract_file, 'r') as f:
        code = f.read()

    patched = False
    if 'np.float(' in code:
        code = code.replace('np.float(', 'float(')
        patched = True
    if 'np.int(' in code:
        code = code.replace('np.int(', 'int(')
        patched = True

    if patched:
        with open(abstract_file, 'w') as f:
            f.write(code)
        print("Successfully patched scikit-video library!")
except Exception as e:
    print(f"Could not patch scikit-video automatically: {e}")

# Dynamically find the path of inference_video.py starting from /content
matches = glob.glob('/content/**/inference_video.py', recursive=True)
if matches:
    script_path = os.path.abspath(matches[0])
    print(f"Found script at: {script_path}")

    # Change working directory to the directory of the script so it can resolve its internal imports
    script_dir = os.path.dirname(script_path)
    %cd {script_dir}

    !python3 "{script_path}" --exp={exp} --video={video_path}
else:
    print("Error: inference_video.py still could not be found. Please run the main Setup cell (Cell 1) above manually.")
```
